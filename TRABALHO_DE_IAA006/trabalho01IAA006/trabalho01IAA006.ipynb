A "amostras" de texto precisa ser "transformada" em <b>padrões</b>

<p>Um padrão é um conjunto de características, geralmente representado por um vetor e um conjunto de padrões no formato de tabela. Onde cada linha é um padrão e as colunas as características e, geralmente, na última coluna a <b>classe</b></p>

```python
import random

pre_padroes = []
for frase in ingles:
  pre_padroes.append( [frase, 'inglês'])

for frase in espanhol:
  pre_padroes.append( [frase, 'espanhol'])

for frase in portugues:
  pre_padroes.append( [frase, 'português'])

random.shuffle(pre_padroes)
print(pre_padroes)
```

<h1>Construção dos atributos</h1>
<p>Esse é o coração desse trabalho e que deverá ser desenvolvido por vocês. Pensem em como podemos "medir" cadas frase/sentença e extrair características que melhorem o resultado do processo de identificação.<p>
<p>Após a criação de cada novo atributo, execute as etapas seguintes e registre as métricas da matriz de confusão. Principalmente acurácia e a precisão.</p>

```python
# a entrada é o vetor pre_padroes e a saída desse passo deverá ser "padrões"
import re

def tamanhoMedioFrases(texto):
  palavras = re.split(r"\s", texto)
  #print(palavras)
  tamanhos = [len(s) for s in palavras if len(s)>0]
  #print(tamanhos)
  soma = 0
  for t in tamanhos:
    soma=soma+t
  return soma / len(tamanhos)


def quantidadeAcentos(texto):
    acentos = "áàâãäéèêëíìîïóòôõöúùûüçÁÀÂÃÄÉÈÊËÍÌÎÏÓÒÔÕÖÚÙÛÜÇ"

    quantidade = 0

    for letra in texto:
        if letra in acentos:
            quantidade += 1

    return quantidade

def quantidadeVogais(texto):
    vogais = "aeiouáàâãéêíóôõúAEIOUÁÀÂÃÉÊÍÓÔÕÚ"

    quantidade = 0

    for letra in texto:
        if letra in vogais:
            quantidade += 1

    return quantidade


def frequenciaPalavra(texto, palavra):
    return len(re.findall(r"\b" + re.escape(palavra.lower()) + r"\b", texto.lower()))

def quantidadeOcorrencias(texto, sequencia):
    return texto.lower().count(sequencia.lower())

def quantidadePalavras(texto):
    return len(texto.split())

def quantidadeConsoantes(texto):
    consoantes = "bcdfghjklmnpqrstvwxyzBCDFGHJKLMNPQRSTVWXYZ"
    return sum(1 for letra in texto if letra in consoantes)

def quantidadeTil(texto):
    return texto.lower().count("ã")

def ocorrenciaCao(texto):
    return texto.lower().count("ção")

def ocorrenciaIng(texto):
    return texto.lower().count("ing")

def ocorrenciaLL(texto):
    return texto.lower().count("ll")

def extraiCaracteristicas(frase):
  # frase é um vetor [ 'texto', 'lingua' ]
  texto = frase[0]
  pattern_regex = re.compile(r'[^\w+]', re.UNICODE)
  texto = re.sub(pattern_regex,' ',texto)
  #print(texto)
  caracteristica1=tamanhoMedioFrases(texto)
  caracteristica2=quantidadeAcentos(texto)
  caracteristica3 = quantidadeVogais(texto)
  caracteristica4 = quantidadeOcorrencias(texto, "th")
  caracteristica5 = quantidadeOcorrencias(texto, "nh")
  caracteristica6 = quantidadeOcorrencias(texto, "lh")
  caracteristica7 = quantidadeOcorrencias(texto, "ch")
  caracteristica8 = quantidadeOcorrencias(texto, "w")
  caracteristica9 = quantidadeOcorrencias(texto, "k")
  caracteristica10 = quantidadeOcorrencias(texto, "y")
  caracteristica11 = quantidadeOcorrencias(texto, "ç")
  caracteristica12 = quantidadeOcorrencias(texto, "ñ")
  caracteristica13=frequenciaPalavra(texto, "the")
  caracteristica14=frequenciaPalavra(texto, "el")
  caracteristica15=frequenciaPalavra(texto, "que")
  caracteristica16=frequenciaPalavra(texto, "and")
  caracteristica17 = frequenciaPalavra(texto, "is")
  caracteristica18 = frequenciaPalavra(texto, "of")
  caracteristica19 = frequenciaPalavra(texto, "la")
  caracteristica20 = quantidadePalavras(texto)
  caracteristica21 = quantidadeConsoantes(texto)
  caracteristica22 = quantidadeTil(texto)
  caracteristica23 = ocorrenciaCao(texto)
  caracteristica24 = ocorrenciaIng(texto)
  caracteristica25 = ocorrenciaLL(texto)
  caracteristica26 = frequenciaPalavra(texto, "não")
  caracteristica27 = frequenciaPalavra(texto, "para")
  caracteristica28 = frequenciaPalavra(texto, "you")
  caracteristica29 = frequenciaPalavra(texto, "to")
  caracteristica30 = frequenciaPalavra(texto, "una")
  caracteristica31 = frequenciaPalavra(texto, "como")

  # acrescente as suas funcoes no vetor padrao
  padrao = [caracteristica1,
    caracteristica2,
    caracteristica3,
    caracteristica4,
    caracteristica5,
    caracteristica6,
    caracteristica7,
    caracteristica8,
    caracteristica9,
    caracteristica10,
    caracteristica11,
    caracteristica12,
    caracteristica13,
    caracteristica14,
    caracteristica15,
    caracteristica16,
    caracteristica17,
    caracteristica18,
    caracteristica19,
    caracteristica20,
    caracteristica21,
    caracteristica22,
    caracteristica23,
    caracteristica24,
    caracteristica25,
    caracteristica26,
    caracteristica27,
    caracteristica28,
    caracteristica29,
    caracteristica30,
    caracteristica31, frase[1] ]
  
  return padrao


def geraPadroes(frases):
  padroes = []
  for frase in frases:
    padrao = extraiCaracteristicas(frase)
    padroes.append(padrao)
  return padroes

# converte o formato [frase classe] em
# [caracteristica_1, caracteristica_2,... caracteristica n, classe]
padroes = geraPadroes(pre_padroes)

#
# apenas para visualizacao
print(padroes)

dados = pd.DataFrame(padroes)
dados
```

<h2>Treinando o modelo com SVM</h1>
<p>Separando o conjunto de treinamento do conjunto de testes</p>

```python
from sklearn.model_selection import train_test_split
import numpy as np
from sklearn.preprocessing import StandardScaler

#from sklearn.metrics import confusion_matrix

vet = np.array(padroes)
classes = vet[:,-1]         # classes = [p[-1] for p in padroes]
#print(classes)
padroes_sem_classe = vet[:,0:-1]
#print(padroes_sem_classe)
X_train, X_test, y_train, y_test = train_test_split(
    padroes_sem_classe,
    classes,
    test_size=0.25,
    stratify=classes,
    random_state=42
)

from sklearn.preprocessing import StandardScaler

# Cria o normalizador
scaler = StandardScaler()

# Aprende a escala usando apenas o conjunto de treino
X_train = scaler.fit_transform(X_train)

# Aplica a mesma escala ao conjunto de teste
X_test = scaler.transform(X_test)
```

Com os conjuntos separados, podemos "treinar" o modelo usando a SVM.

```python
from sklearn import svm
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report

treinador = svm.SVC()  #algoritmo escolhido
modelo = treinador.fit(X_train, y_train)

#
# score com os dados de treinamento
acuracia = modelo.score(X_train, y_train)
print("Acurácia nos dados de treinamento: {:.2f}%".format(acuracia * 100))

#
# melhor avaliar com a matriz de confusão
y_pred = modelo.predict(X_train)
cm = confusion_matrix(y_train, y_pred)
print(cm)
print(classification_report(y_train, y_pred))

#
# com dados de teste que não foram usados no treinamento
print('métricas mais confiáveis')
y_pred2 = modelo.predict(X_test)
cm = confusion_matrix(y_test, y_pred2)
print(cm)
print(classification_report(y_test, y_pred2))
```
