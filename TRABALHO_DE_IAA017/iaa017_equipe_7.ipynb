### Validando dados

````python
# Criando categorias para separar numéricas e categóricas
numericas_cols = [col for col in dados['Sales'].columns if dados['Sales'][col].dtype != 'str']
categoricas_cols = [col for col in dados['Sales'].columns if dados['Sales'][col].dtype == 'str']
````

````python
# Exibindo as estatísticas descritivas para as colunas numéricas da planilha 'Sales'
dados['Sales'][numericas_cols].describe()
````

````python
# Exibindo as estatísticas descritivas para as colunas categóricas da planilha 'Sales'
dados['Sales'][categoricas_cols].describe()
````
